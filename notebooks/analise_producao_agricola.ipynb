```python
!pip install pandas plotly

import pandas as pd
import plotly.express as px

df = pd.read_csv('simulacao_producao_agricola_brasil.csv')
df['ano'] = df['ano'].astype(int)

df = df.drop_duplicates()

print("--- VISUALIZAÇÃO INICIAL DOS DADOS ---")
display(df.head())

print("\n--- VERIFICAÇÃO DE VALORES NULOS ---")
print(df.isnull().sum())
```

```python
prod_total = df['producao_toneladas'].sum()
valor_total = df['valor_producao'].sum()
area_total = df['area_plantada_ha'].sum()
cultura_top = df.groupby('cultura')['producao_toneladas'].sum().idxmax()
estado_top = df.groupby('uf')['producao_toneladas'].sum().idxmax()
media_produtividade = df['produtividade'].mean()

print("==========================================")
print(f" INDICADORES DE DESEMPENHO (KPIs)")
print("==========================================")
print(f"• Produção Total: {prod_total:,.0f} toneladas")
print(f"• Valor Econômico Total: R$ {valor_total:,.2f}")
print(f"• Área Plantada Total: {area_total:,.0f} ha")
print(f"• Cultura Mais Produzida: {cultura_top}")
print(f"• Estado Mais Produtivo: {estado_top}")
print(f"• Média de Produtividade: {media_produtividade:.2f} t/ha")
print("==========================================")
```

```python
fig_tempo = px.line(
    df.groupby('ano')['producao_toneladas'].sum().reset_index(),
    x='ano', y='producao_toneladas',
    title='Evolução Temporal da Produção Agrícola (2015-2024)',
    markers=True
)
fig_tempo.show()

fig_estado = px.bar(
    df.groupby('uf')['producao_toneladas'].sum().reset_index(),
    x='uf', y='producao_toneladas',
    title='Produção Total por Estado (UF)',
    color='uf'
)
fig_estado.show()

fig_clima = px.scatter(
    df, x='chuva_mm', y='produtividade', color='cultura',
    title='Correlação: Volume de Chuvas (mm) x Produtividade (t/ha)'
)
fig_clima.show()
```
